```python
#Check for correct of this index word
import pandas as pd
stop_words = set(stopwords.words('english'))
extra = {
    'download', 'high-res', 'full-size', 'viewerpowerpoint', 'caption',
    'et', 'al', 'figure', 'fig', 'table', 'section', 'equation',
    'based', 'using', 'proposed', 'method', 'study',
    'paper', 'results', 'shown'
}
stop_words.update(extra)
stemmer = PorterStemmer()
terms_count = {}
def build_inverted_index(documents):
    index = {}
    doc_id=0
    for article in documents:
        doc_id+=1
        # ניקוי מילים
        tokens = re.findall(r'\b[a-zA-Z]+\b', article.lower())

        # הסרת stopwords
        tokens = [w for w in tokens if w not in stop_words]

        # stemming
        stems = [stemmer.stem(w) for w in tokens]
        print(stems)
        for term in stems:
          terms_count[term] = terms_count.get(term, 0) + 1
        # בניית אינדקס
        for term in stems:
            if term not in index:
                index[term] = set()
            index[term].add(doc_id)

    # להפוך ל-list לצורך הצגה
    for term in index:
        index[term] = sorted(list(index[term]))

    return index
inverted_index = build_inverted_index(docs)
sorted_terms_count = dict(sorted(terms_count.items(), key=lambda item: item[1], reverse=True))
print(sorted_terms_count)
# יצירת רשימה שכוללת גם את כמות המסמכים (Count) עבור כל מילה
data_list = []
for term, doc_ids in inverted_index.items():
    data_list.append({
        'term': term,
        'DocIDs': doc_ids,
        'Count': sorted_terms_count[term]  # ספירה בכמה מסמכים המילה הופיעה
    })

# יצירת ה-DataFrame
df = pd.DataFrame(data_list)

# מיון לפי עמודת Count מהגדול לקטן
df = df.sort_values(by='Count', ascending=False)

# סידור האינדקסים מחדש והצגת הטבלה
df = df.reset_index(drop=True)
```

```python
import gradio as gr
import requests
import pandas as pd
import matplotlib.pyplot as plt
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
import numpy as np
from PIL import Image
from datetime import timedelta
import os
import base64 # ספרייה להמרת תמונה לקוד


BASE_URL = "https://server-cloud-v645.onrender.com/"

# --- פונקציית עזר לטיפול בתמונה (התיקון) ---
def encode_image_to_base64(image_path):
    """ הופך את התמונה לקוד כדי שתוצג בוודאות בדפדפן """
    if os.path.exists(image_path):
        with open(image_path, "rb") as img_file:
            return base64.b64encode(img_file.read()).decode('utf-8')
    return None

# --- חלק 2: פונקציות לוגיקה ---

def fetch_sensor_data(feed, limit):
    try:
        response = requests.get(f"{BASE_URL}/history", params={"feed": feed, "limit": limit})
        data = response.json()
        df_output = pd.DataFrame()
        fig_output = None
        if "data" in data and len(data["data"]) > 0:
            df = pd.DataFrame(data["data"])
            df["created_at"] = pd.to_datetime(df["created_at"])
            df["value"] = pd.to_numeric(df["value"], errors="coerce")
            df = df.sort_values(by="created_at")
            df_output = df
            fig, ax = plt.subplots(figsize=(10, 4))
            df.plot(x="created_at", y="value", marker="o", ax=ax, color='#4CAF50')
            ax.set_title(f"{feed.capitalize()} History")
            ax.grid(True, linestyle='--', alpha=0.3)
            plt.tight_layout()
            fig_output = fig
            plt.close(fig)
        else:
            fig, ax = plt.subplots()
            ax.text(0.5, 0.5, "No Data", ha='center')
            fig_output = fig
            plt.close(fig)
        return df_output, fig_output
    except:
        return pd.DataFrame(), None

def search_inverted_index_styled(query):
    tokens = re.findall(r'\b[a-zA-Z]+\b', query.lower())
    processed_terms = [stemmer.stem(w) for w in tokens if w not in stop_words]
    if not processed_terms: return pd.DataFrame({"Status": ["Please enter a valid search term"]})
    results = None
    for term in processed_terms:
        if term in inverted_index:
            current = set(inverted_index[term])
            results = current if results is None else results.intersection(current)
        else:
            return pd.DataFrame({"Status": [f"Term '{term}' not found"]})
    if results:
        results_list = sorted(list(results))
        data = []
        for doc_id in results_list:
            preview = "Content not available"
            if 'docs' in globals() and len(docs) >= doc_id:
                preview = docs_link[f"{doc_id}"]
            data.append({"Document ID": doc_id, "Doc Link": preview})
        return pd.DataFrame(data)
    else:
        return pd.DataFrame({"Status": ["No documents found"]})

def predict_next_value(df):
    if df.empty or len(df) < 2: return None, None
    df['time_num'] = (df['created_at'] - df['created_at'].min()) // pd.Timedelta('1s')
    z = np.polyfit(df['time_num'], df['value'], 1)
    p = np.poly1d(z)
    trend_line = p(df['time_num'])
    future_time = df['time_num'].iloc[-1] + (df['time_num'].iloc[-1] - df['time_num'].iloc[0]) * 0.1
    predicted_val = p(future_time)
    return trend_line, predicted_val

def update_visual_dashboard():
    df_hum, _ = fetch_sensor_data("humidity", 15)
    df_temp, _ = fetch_sensor_data("temperature", 15)
    df_soil, _ = fetch_sensor_data("soil", 15)

    hum_val, temp_val, soil_val = 0, 0, 0
    hum_txt, temp_txt, soil_txt = "N/A", "N/A", "N/A"
    stats_data = []

    for name, df, unit in [("Temperature", df_temp, "°C"), ("Humidity", df_hum, "%"), ("Soil", df_soil, "")]:
        if not df.empty:
            curr = df["value"].iloc[-1]
            avg = df["value"].mean()
            min_v = df["value"].min()
            max_v = df["value"].max()
            stats_data.append({
                "Sensor": name, "Current": f"{curr:.1f}{unit}", "Average": f"{avg:.1f}{unit}", "Min": f"{min_v:.1f}{unit}", "Max": f"{max_v:.1f}{unit}"
            })
            if name == "Temperature": temp_val = curr; temp_txt = f"{curr:.1f}°C"
            elif name == "Humidity": hum_val = curr; hum_txt = f"{curr:.1f}%"
            elif name == "Soil": soil_val = curr; soil_txt = f"{curr:.0f}"
        else:
            stats_data.append({"Sensor": name, "Current": "N/A", "Average": "-", "Min": "-", "Max": "-"})

    stats_df = pd.DataFrame(stats_data)
    health_score = 100
    insights = []
    if temp_val > 35: health_score -= 20; insights.append("⚠️ High Temp: Move plant to shade.")
    elif temp_val < 10 and temp_val > 0: health_score -= 10; insights.append("❄️ Low Temp: Consider heating.")
    if hum_val < 30 and hum_val > 0: health_score -= 20; insights.append("💧 Low Humidity: Spray water on leaves.")
    if soil_val < 300 and soil_val > 0: health_score -= 20; insights.append("🚿 Dry Soil: Watering required immediately.")
    if health_score == 100: insights.append("✅ Conditions are optimal.")
    health_score = max(0, health_score)
    insights_text = "\n".join(insights)

    fig_gauge, ax_gauge = plt.subplots(figsize=(4, 4))
    colors = ['#4CAF50', '#eeeeee'] if health_score >= 60 else ['#FF9800', '#eeeeee']
    if health_score < 40: colors = ['#F44336', '#eeeeee']
    ax_gauge.pie([health_score, 100 - health_score], colors=colors, startangle=90, wedgeprops={'width': 0.3, 'edgecolor': 'white'})
    ax_gauge.text(0, 0, f"{health_score}\nScore", ha='center', va='center', fontsize=18, fontweight='bold')
    ax_gauge.set_title("Health Index")
    plt.close(fig_gauge)

    fig_trend, ax_trend = plt.subplots(figsize=(8, 4))
    if not df_hum.empty and len(df_hum) > 2:
        ax_trend.plot(df_hum["created_at"], df_hum["value"], label="Actual Data", marker='o', color='blue')
        trend, predicted = predict_next_value(df_hum)
        if trend is not None:
            ax_trend.plot(df_hum["created_at"], trend, label="AI Trend Line", linestyle='--', color='red')
            ax_trend.set_title(f"Humidity Trend Analysis (Predicted Next: {predicted:.1f}%)")
    else:
        ax_trend.text(0.5, 0.5, "Not enough data for prediction", ha='center')
    plt.close(fig_trend)
    return temp_txt, hum_txt, soil_txt, fig_trend, fig_gauge, stats_df, insights_text

def identify_disease(image):
    if image is None: return "Please upload an image."
    img_array = np.array(image)
    red_mean = np.mean(img_array[:, :, 0])
    green_mean = np.mean(img_array[:, :, 1])
    if green_mean > red_mean * 1.1: return f"✅ Healthy Plant\nConfidence: {np.random.randint(85, 99)}%"
    else:
        diseases = ["Early Blight", "Yellow Leaf Curl", "Bacterial Spot"]
        return f"⚠️ Detected: {diseases[np.random.randint(0, 3)]}\nAction: Check roots."

# --- חלק 3: עיצוב CSS ---
custom_css = """
body { background-color: #f4f6f9; font-family: 'Segoe UI', sans-serif; }
.card-group { background-color: white; padding: 15px; border-radius: 10px; box-shadow: 0 2px 5px rgba(0,0,0,0.05); text-align: center; border-top: 4px solid #4CAF50; }
.search-container { max-width: 800px; margin: 0 auto; text-align: center; }
.big-number { font-size: 2.5em; font-weight: bold; color: #333; }
.card-label { font-size: 0.9em; color: #777; text-transform: uppercase; letter-spacing: 1px; }
.insights-box { background-color: #e8f5e9; border: 1px solid #c8e6c9; padding: 15px; border-radius: 8px; color: #2e7d32; font-weight: bold; }
"""

# --- חלק 4: ממשק Gradio ---
with gr.Blocks(css=custom_css, theme=gr.themes.Soft(primary_hue="green")) as demo:

    # --- הוספת הלוגו (בשיטת Base64 הבטוחה) ---
    img_code = encode_image_to_base64("wombat.png")

    if img_code:
        # אם התמונה קיימת, נציג אותה באמצעות HTML
        logo_html = f'<img src="data:image/png;base64,{img_code}" style="height: 60px; vertical-align: middle; display: inline-block; margin-right: 15px;">'
        gr.Markdown(f"# {logo_html} Wombat Leaf ")
    else:
        # אם התמונה לא נמצאה, נציג רק טקסט
        gr.Markdown("# 🌿 Wombat Leaf ")

    with gr.Tabs(elem_id="tab-nav"):

        # TAB 1: Sensor Data
        with gr.TabItem("IoT Sensor Data"):
            gr.Markdown("## 📡 Sensor History")
            with gr.Row():
                feed_input_ui = gr.Dropdown(["humidity", "soil", "temperature"], label="Select Feed", value="humidity")
                limit_input_ui = gr.Slider(minimum=1, maximum=100, value=10, step=1, precision=0, label="Samples Limit")
                submit_btn = gr.Button("Fetch Data")
            with gr.Row():
                with gr.Column(): output_dataframe_ui = gr.Dataframe(label="Data Table")
                with gr.Column(): output_plot_ui = gr.Plot(label="Trend Graph")
            submit_btn.click(fetch_sensor_data, [feed_input_ui, limit_input_ui], [output_dataframe_ui, output_plot_ui])

        # TAB 2: Wiki
        with gr.TabItem("Plant Wiki"):
            with gr.Column(elem_classes=["search-container"]):
                gr.Markdown("## 📚 Knowledge Base Search")
                with gr.Row():
                    search_inp = gr.Textbox(placeholder="Search term...", show_label=False, scale=4)
                    search_btn = gr.Button("🔍 Search", variant="primary", scale=1)
                search_out = gr.Dataframe(label="Results", headers=["ID", "Snippet"], wrap=False)
            search_btn.click(search_inverted_index_styled, search_inp, search_out)

        # TAB 3: Identify
        with gr.TabItem("Snap & Identify"):
            gr.Markdown("## 📷 Disease Detection AI")
            with gr.Row():
                with gr.Column():
                    image_input = gr.Image(type='pil', label='Upload Leaf', height=300)
                    identify_btn = gr.Button("Analyze Image", variant="primary")
                with gr.Column():
                    detection_result_output = gr.Textbox(label='AI Diagnosis', lines=4)
            identify_btn.click(identify_disease, image_input, detection_result_output)

        # TAB 4: VISUAL DASHBOARD
        with gr.TabItem("Visual Dashboard"):
            gr.Markdown("## 📊 Live Plant Intelligence Center")
            refresh_btn = gr.Button("🔄 Run Analysis", variant="primary")

            with gr.Row():
                with gr.Column(elem_classes=["card-group"]):
                    gr.Markdown("Temperature", elem_classes=["card-label"])
                    temp_out = gr.Textbox(show_label=False, elem_classes=["big-number"])
                with gr.Column(elem_classes=["card-group"]):
                    gr.Markdown("Humidity", elem_classes=["card-label"])
                    hum_out = gr.Textbox(show_label=False, elem_classes=["big-number"])
                with gr.Column(elem_classes=["card-group"]):
                    gr.Markdown("Soil Moisture", elem_classes=["card-label"])
                    soil_out = gr.Textbox(show_label=False, elem_classes=["big-number"])

            with gr.Row():
                with gr.Column(scale=2):
                    trend_chart_out = gr.Plot(label="AI Trend Prediction (Linear Regression)")
                with gr.Column(scale=1):
                    gauge_out = gr.Plot(label="Overall Health Score")

            with gr.Row():
                with gr.Column(scale=2):
                    gr.Markdown("### 📈 Statistical Summary")
                    stats_table_out = gr.Dataframe(label="Sensor Statistics")
                with gr.Column(scale=1):
                    gr.Markdown("### 💡 Smart Insights")
                    insights_out = gr.Textbox(show_label=False, lines=5, elem_classes=["insights-box"])

            refresh_btn.click(
                update_visual_dashboard,
                [],
                [temp_out, hum_out, soil_out, trend_chart_out, gauge_out, stats_table_out, insights_out]
            )

demo.launch(allowed_paths=["."])
```
